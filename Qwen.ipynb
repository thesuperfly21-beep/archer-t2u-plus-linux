```python
import torch
from diffusers import QwenImageEditPipeline
from diffusers.utils import load_image

# Initialize the Qwen Image Edit pipeline in bfloat16 to optimize VRAM
model_id = "Qwen/Qwen-Image-Edit"
pipe = QwenImageEditPipeline.from_pretrained(model_id, torch_dtype=torch.bfloat16)
pipe.to("cuda")

# 1. Load your input image (Replace this URL with your own image URL or local path)
input_url = "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/diffusers/cat.png"
input_image = load_image(input_url).convert("RGB")

# 2. Define the editing prompt instruction
prompt = "Turn this cat into a realistic dog, highly detailed"

# 3. Generate the edited image
print("Processing image edit...")
edited_image = pipe(image=input_image, prompt=prompt).images[0]

# 4. Display the result in Colab
display(edited_image)
```
